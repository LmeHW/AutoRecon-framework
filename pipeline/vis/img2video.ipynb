```python
from pathlib import Path
import cv2

# img_dir = Path("/mnt/sda/public/00_Projects/2512_AutoRecon/data/video/3tfei/Axial")   # 改成你的图片文件夹
# out_mp4 = "/mnt/sda/public/00_Projects/2512_AutoRecon/data/video/3tfei/axial.mp4"
img_dir = Path("/mnt/sda/public/00_Projects/2512_AutoRecon/data/video/15baoshu/axial")   # 改成你的图片文件夹
out_mp4 = "/mnt/sda/public/00_Projects/2512_AutoRecon/data/video/15baoshu/axial.mp4"
# img_dir = Path("/mnt/sda/public/00_Projects/2512_AutoRecon/data/video/15baoshu/sag")   # 改成你的图片文件夹
# out_mp4 = "/mnt/sda/public/00_Projects/2512_AutoRecon/data/video/15baoshu/sag.mp4"
img_dir = Path("/mnt/sda/public/00_Projects/2512_AutoRecon/data/video/15baoshu/cor")   # 改成你的图片文件夹
out_mp4 = "/mnt/sda/public/00_Projects/2512_AutoRecon/data/video/15baoshu/cor.mp4"
fps = 12

exts = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
paths = sorted([p for p in img_dir.iterdir() if p.suffix.lower() in exts])

img0 = cv2.imread(str(paths[0]))
h, w = img0.shape[:2]
fourcc = cv2.VideoWriter_fourcc(*"mp4v")
vw = cv2.VideoWriter(out_mp4, fourcc, fps, (w, h))

for p in paths:
    im = cv2.imread(str(p))
    if im.shape[:2] != (h, w):
        im = cv2.resize(im, (w, h))
    vw.write(im)

vw.release()
print("saved:", out_mp4, "frames:", len(paths))
```
