```python
import argparse, math
from pathlib import Path
import ants
import pandas as pd
```

```python
def nearest_week(ga, wk_min=22, wk_max=35):
    return max(wk_min, min(wk_max, int(math.floor(ga + 0.5))))


def reg(img_names, ga):
    wk = nearest_week(ga)
    atlas_img = f"/mnt/sda/public/00_Projects/2512_AutoRecon/data/FBA_Chinese_Atlas/img/{wk:02d}w.nii.gz"
    atlas_roi = f"/mnt/sda/public/00_Projects/2512_AutoRecon/data/FBA_Chinese_Atlas/ROI/{wk:02d}w.nii.gz"
    subject = f"/mnt/sda/public/00_Shares/Fetal_brain/gt/{img_names}.nii.gz"
    out_roi=f"/mnt/sda/public/00_Projects/2512_AutoRecon/data/gfy_ROI/{img_names}.nii.gz"
 

    fixed = ants.image_read(str(subject))      # subject space
    moving = ants.image_read(str(atlas_img))   # atlas space

    reg = ants.registration(fixed=fixed, moving=moving, type_of_transform="SyN")

    roi = ants.image_read(str(atlas_roi))
    roi_warp = ants.apply_transforms(
        fixed=fixed, moving=roi,
        transformlist=reg["fwdtransforms"],
        interpolator="nearestNeighbor"
    )

    ants.image_write(roi_warp, out_roi)
    print(f"[OK] GA={ga:.2f} -> wk={wk}, wrote: {out_roi}")

```
